```python
full_name = "Bilal"
parts = full_name.split()
first_name = parts[0]
last_name = parts[-1]

# f-string
print(f"My first name is {first_name} and my last name is {last_name}.")

# Count vowels
vowel_count = 0
for char in full_name.lower():
    if char in "aeiou":
        vowel_count += 1
print("Number of vowels in full name:", vowel_count)

# Uppercase vs lowercase comparison
upper_name = full_name.upper()
lower_name = full_name.lower()
print("Uppercase:", upper_name)
print("Lowercase:", lower_name)
print("Uppercase equals lowercase?", upper_name == lower_name)
```

```python
marks = 85
if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"
print("Marks:", marks, "Grade:", grade)


number = -5
if number > 0:
    print("Number is positive")
elif number < 0:
    print("Number is negative")
else:
    print("Number is zero")

year = 2024
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year")


cgpa = 3.8
attendance = 85
eligible = cgpa > 3.5 and attendance > 80
print("Scholarship eligible:", eligible)
```

Task7

```python
print("Numbers 1-30 skipping multiples of 4:")
for i in range(1, 31):
    if i % 4 == 0:
        continue
    print(i, end=" ")
print()

print("Numbers 1-50 stopping after 20:")
for i in range(1, 51):
    if i > 20:
        break
    print(i, end=" ")
print()

num = 7
i = 1
print("Multiplication table of", num)
while i <= 10:
    print(f"{num} x {i} = {num * i}")
    i += 1

total = 0
for i in range(1, 101):
    if i % 2 == 0:
        total += i
print("Sum of even numbers from 1 to 100:", total)
```

```python
def calculate_area(length, width=1):
    return length * width


print("Area with both arguments:", calculate_area(5, 3))
print("Area with only length:", calculate_area(5))
area_result = calculate_area(4, 2)
print("Area stored in variable:", area_result)


def is_even(number):
    return number % 2 == 0

print("Is 10 even?", is_even(10))
print("Is 7 even?", is_even(7))

#  List passed by reference
def add_item(my_list, item):
    print("Inside function before adding:", my_list)
    my_list.append(item)
    print("Inside function after adding:", my_list)

sample_list = ["apple", "banana"]
print("Outside before function call:", sample_list)
add_item(sample_list, "cherry")
print("Outside after function call:", sample_list)
```

Task9

```python
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours

    def display_info(self):
        print(f"Course: {self.course_name}, Credit Hours: {self.credit_hours}")


course1 = Course("AIC380", 3)
course2 = Course("CSC101", 4)
course1.display_info()
course2.display_info()


class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

    # Override
    def display_info(self):
        super().display_info()
        print(f"Lab Hours: {self.lab_hours}")

lab_course = LabCourse("Robotics", 3, 2)
lab_course.display_info()
```
