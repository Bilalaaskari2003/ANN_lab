Task3

```python
import numpy as np
```

```python

term1 = np.array([
    [70, 75, 80],
    [60, 65, 70],
    [85, 80, 90]
])


term2 = np.array([
    [80, 80, 85],
    [70, 75, 80],
    [90, 85, 95]
])
```

```python
combined_marks = term1 + term2

print("Combined Marks:")
print(combined_marks)

improvement = term2 - term1

print("Improvement:")
print(improvement)
```

```python
total_improvement = np.sum(improvement,axis=1)

print("Total Improvement of each student:")
print(total_improvement)
```

```python
highest_student = np.argmax(total_improvement)

print("Student with highest total improvement:",
      highest_student + 1)
```

```python
import numpy as np
```

```python
prices = np.array([100, 200, 150, 50])
quantities = np.array([2, 3, 1, 4])
```

```python
#total bill using dot product
total_bill = np.dot(prices, quantities)

print("Total bill for Customer 1:", total_bill)
```

```python
#Quantities purchased by 2 different customers
customer_quantities = np.array([
    [2, 3, 1, 4],
    [1, 2, 3, 2]
])
```

```python
#total bill for both customers using a single dot product
total_bills = np.dot(customer_quantities, prices)
print("Total bills for both customers:", total_bills)
# Explanation:
# customer_quantities has shape (2, 4) and prices has shape (4,).
# The inner dimensions must match: 4 == 4.
# This allows NumPy to multiply each customer's 4 quantities
# with the corresponding 4 prices and calculate both totals.
```

```python
import numpy as np
performance = np.array([
    [80, 85, 90, 88, 92],
    [70, 75, 78, 82, 85],
    [90, 92, 95, 94, 96]
])

print("Original Matrix:")
print(performance)
```

```python
# Transpose the matrix
transposed = performance.T

print("\nTransposed Matrix:")
print(transposed)

# Print the new shape
print("Original Shape:", performance.shape)
print("Transposed Shape:", transposed.shape)
```

```python
# Verify programmatically that transposing twice gives the original matrix
double_transposed = performance.T.T

print("Does transposing twice give the original matrix?",
      np.array_equal(performance, double_transposed))
```

```python
# Create a 4x2 matrix
matrix_a = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

# Create a 3x2 matrix
matrix_b = np.array([
    [10, 20],
    [30, 40],
    [50, 60]
])

# Transpose matrix_b from 3x2 to 2x3
matrix_b_transposed = matrix_b.T

print("Matrix A Shape:", matrix_a.shape)
print("Matrix B Shape:", matrix_b.shape)
print("Transposed Matrix B Shape:", matrix_b_transposed.shape)
```

```python
result = np.dot(matrix_a, matrix_b_transposed)

print("Dot Product Result:")
print(result)
```
